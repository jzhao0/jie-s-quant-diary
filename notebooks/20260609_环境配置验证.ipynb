```python
# 获取A股数据示例
df = ak.stock_zh_a_hist(symbol="600519", period="daily", start_date="20260101", end_date="20260609")
df.head()
```

```python
# 计算移动平均线
df['MA5'] = talib.SMA(df['收盘'].values, timeperiod=5)
df['MA20'] = talib.SMA(df['收盘'].values, timeperiod=20)

# 可视化
plt.figure(figsize=(12, 6))
plt.plot(df['日期'], df['收盘'], label='收盘价')
plt.plot(df['日期'], df['MA5'], label='MA5')
plt.plot(df['日期'], df['MA20'], label='MA20')
plt.legend()
plt.title('贵州茅台 (600519) 股价走势')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
